```python
from functions import read_img,show_img,preproces_pipeline
import cv2
import numpy as np
from skimage.filters import frangi,sato
import torch

DEVICE:str="cuda" if torch.cuda.is_available() else "cpu"
print(f"USED DEVICE: {DEVICE}")
```

```python
img=read_img(".\images\\01_dr.JPG")
show_img(img[:,:,1],"Before processing")
img=preproces_pipeline(img)
show_img(img,"After processing")
vesesl=sato(img,[1,2,3,4])
vesesl=np.where(vesesl>0.03,1,0)
show_img(vesesl,"vessels no clip")

img=read_img(".\images\\01_dr.JPG")
img=preproces_pipeline(img,clip=True)
show_img(img,"After processing with clip")
vesesl=sato(img,[1,2,3,4])
print(vesesl.max())
vesesl=np.where(vesesl>0.03,1,0)
show_img(vesesl,"vessels clip")
```
